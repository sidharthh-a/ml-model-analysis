```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Generating synthetic data
np.random.seed(42)
X = np.random.rand(100, 1) * 10    # 100 random numbers between 0 and 10
y = 3 * X + 7 + np.random.randn(100, 1) * 2  # true relation + noise
```

```python
# Initialize m and c (slope and intercept)
m = 0.0
c = 0.0

# Learning rate and iterations
learning_rate = 0.01
epochs = 1000

# To store the loss for visualization
loss_history = []
```

```python
# Gradient Descent loop
for i in range(epochs):
    y_pred = m * X + c                   # Predictions
    error = y_pred - y                   # Error
    cost = np.mean(error ** 2)           # Mean Squared Error
    loss_history.append(cost)

    # Gradients
    dm = (2 / len(X)) * np.sum(error * X)
    dc = (2 / len(X)) * np.sum(error)

    # Update parameters
    m -= learning_rate * dm
    c -= learning_rate * dc

    # Print progress occasionally
    if i % 100 == 0:
        print(f"Epoch {i}: Loss={cost:.4f}, m={m:.4f}, c={c:.4f}")
```

```python
# Plot loss curve
plt.plot(loss_history)
plt.title("Loss vs Epochs")
plt.xlabel("Epochs")
plt.ylabel("Mean Squared Error")
plt.show()

# Plot regression line
plt.scatter(X, y, color='blue', label='Data Points')
plt.plot(X, m * X + c, color='red', label='Fitted Line')
plt.title("Linear Regression using Gradient Descent")
plt.legend()
plt.show()

print(f"Final parameters: m = {m:.3f}, c = {c:.3f}")
```
